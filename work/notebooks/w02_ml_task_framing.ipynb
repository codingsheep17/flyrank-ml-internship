## 3. Success metric

*One metric you can defend. What number means 'good'?*

In the starter pipeline, this metric showed a hand-written rule scoring 0.240 versus a Random Forest scoring 0.740 — a clear, measurable improvement.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule can only check a few conditions someone thought of in advance (e.g. "flag pages with low CTR and old age"). But real decline patterns involve many signals interacting together — position, age, word count, engagement, freshness — in ways a human can't hand-write as a simple if/else rule. This is exactly what the starter pipeline already proved: a hand-written rule scored 0.240 on Precision@50, while a Random Forest trained on the same data scored 0.740 — roughly 3x better at picking real declining pages into the top 50.
